# Breast Cancer Wisconsin Diagnostic Dataset

```python
import pandas as pd 
import numpy as np
import sklearn 
import matplotlib.pyplot as plt
from IPython.display import display
from sklearn.model_selection import train_test_split
```

## Load the data

```python
from sklearn.datasets import load_breast_cancer
cancer_data = load_breast_cancer()
X = pd.DataFrame(
    cancer_data.data, 
    columns=cancer_data.feature_names) 
y = pd.Series(cancer_data.target)
```

```python
# Display first five features
print("Features: ")
display(X.head())

# Display the first 5 target values and their name
print("Target y and Target Names:")
print(y.head())
print("Target Names:", cancer_data.target_names)
```

## Split the data

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle= True, 
                                                    stratify=y, random_state= 42)

print("Shape of X_train: ", X_train.shape)
print("Shape of X_test: ", X_test.shape)
print("Shape of y_train: ", y_train.shape)
print("Shape of y_test: ", y_test.shape)
```

## Preprocess Data

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
# Fit the scaler and transform on both training and testing data
# FIT only on TRAIN SET to prevent lakage data from the test set 
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)


X_train_scaled = pd.DataFrame(
    scaler.fit_transform(X_train),
    columns=X_train.columns,
    index=X_train.index
)

X_test_scaled = pd.DataFrame(
    scaler.transform(X_test),
    columns=X_test.columns,
    index=X_test.index
)

print("First 5 rows in original training set: ")
display(X_train.head(5))
print("First 5 rows in scaled training set: ")
display(X_train_scaled.head(5))
```

## Choose and train the model

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression(solver='liblinear', random_state= 42)
print("Chosen model: ", model)
```

```python
# Train the model
# Fit the model on training data
model.fit(X_train_scaled, y_train)
print("Model train successfully")
```

```python
# Make decision 
y_pred = model.predict(X_test_scaled)
print("Prediction on the test set: ")
print(y_pred)
print("First 10 actual test: ")
print(y_test.head(10))
print("First 10 actual predict: ")
print(y_pred[:10])
```

## Evaluated the model

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
# Accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy:.6f}")
# Generated the classification report (precision, f1-score per class)
class_report = classification_report(y_test, y_pred, 
                                     target_names=cancer_data.target_names)
print("Classification Report:\n")
print(class_report)
# Confusion Matrix
conf_matrix = confusion_matrix(y_test, y_pred)
print("Confusion Matrix")
print("Rows: True Lables")
print("Columns: Predict Lables")
print(pd.DataFrame(conf_matrix, 
                   index=cancer_data.target_names, 
                   columns=cancer_data.target_names))
```

## Tune Hyperparameters

```python
# Can be use Grid Search or Random Search 
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import make_scorer, recall_score

# Define parameter grid to search 
# small C -> strong penalty
# large C -> weak penalty 
param_grid = {
    'C': [0.001, 0.01, 0.1, 1, 10, 100], 
    'solver': ['liblinear', 'lbfgs']
    }

grid_search = GridSearchCV(LogisticRegression(random_state= 42), 
                           param_grid, cv= 5, 
                           scoring=make_scorer(recall_score, pos_label=0))
grid_search.fit(X_train_scaled, y_train)

print("Best parameters found:", grid_search.best_params_)
print(f"Best Cross - Validation Accuracy: {grid_search.best_score_:.6f}")
# Get the best model 
best_model = grid_search.best_estimator_
# Evaluate the best model on the test set
y_pred_tuned = best_model.predict(X_test_scaled)
accuracy_tuned = accuracy_score(y_test, y_pred_tuned)
print(f"Accuracy with Tuned Model on the test set: {accuracy_tuned:.4f}")

# Get Clasification Report on Tuned Model
print("Classification Report on Tuned Model")
print(classification_report(y_test, y_pred_tuned, 
                            target_names= cancer_data.target_names))

```
